2. Estadistica

Con la funcion describe podemos saber el promedio, la desviacion estandar, quartiles y sus valores minimo y maximo 

```python
iris.describe()
```

Valor promedio del sepalo 

```python
promedio =iris['sepal_length'].mean()
print("valor promedio",promedio)
```

Especie que tiene mayor promedio de petalo

```python
pro= iris.groupby('species')['petal_length'].mean().nlargest(1)
print("valor promedio", pro)
```

```python
Especies y sus promedios 
```

```python
iris.groupby('species').mean()
```

3. Filtrado

Las flores cuya sepal_length sea mayor que 6

```python
iris[iris['sepal_length'] > 6]
```

```python
Flores cuyo sepal_length es mayor que 6 y su petal_length mayor que 5
```

```python
iris[iris['sepal_length']>6]
iris[iris['petal_length']>5]
```

Cuentas flores pertenecen a cada especie 

```python
iris.groupby('species').count()
```

¿Cuántas flores tienen un petal_width mayor que 1.5?

```python
iris[iris['petal_width']>1.5].groupby('species').count()
```

4. Análisis 

¿Cuál es la especie con mayor promedio de petal_width?

```python
iris.groupby("species")["petal_width"].mean().nlargest(1)
```

```python
La mediana de sepal_length para cada especie
```

```python
iris.groupby("species")["sepal_length"].median()
```

Desviación estándar de petal_length para cada especie

```python
iris.groupby("species")["petal_length"].std()
```

```python
la virginica tiene mayor variabilidad 
```

5. Visualización 

Grafica de dispersción

```python
sns.scatterplot(data = iris, x='sepal_length',y='petal_length', hue='species', palette='inferno')
```

Se puede visualizar que entre mas largo el petalo mas largo sera el sepalo 

Las setosa son las flores más pequeñas y las verginica son las flores mas grande seguida de casi del mismo tamaño las flores versicolor
